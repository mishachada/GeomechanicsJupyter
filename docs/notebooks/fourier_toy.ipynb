# Fourier Toy Notebook

This quick exploration shows how partial sums of a Fourier series approximate a square wave. The \(n\)-term approximation is

$$S_n(x) = rac{4}{\pi} \sum_{k=1,3,5,\dots}^{n} rac{\sin(kx)}{k}. $$

Use the code cell below to visualize the approximation for different numbers of terms.


```

import numpy as np
import matplotlib.pyplot as plt

terms = 9
x = np.linspace(-np.pi, np.pi, 500)
series = np.zeros_like(x)

for k in range(1, terms + 1, 2):
    series += np.sin(k * x) / k

approximation = (4 / np.pi) * series

fig, ax = plt.subplots(figsize=(6, 3))
ax.plot(x, approximation, label=f"{terms} terms")
ax.set_ylim(-1.5, 1.5)
ax.set_xlabel("x")
ax.set_ylabel("S_n(x)")
ax.set_title("Square Wave Approximation")
ax.grid(True, alpha=0.3)
ax.legend()
fig
```
